```python
from anthropic import Anthropic

client = Anthropic(
        base_url="https://openrouter.ai/api/"
)
model = "claude-haiku-5.5"
```

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages= [
    {
        "role": "user",
        "content": "What is Kajv Solutions or KAJV.site in Costa Rica"
    }
    ]
)
```
