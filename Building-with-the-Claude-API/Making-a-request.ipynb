```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from anthropic import Anthropic

client = Anthropic(
        base_url="https://openrouter.ai/api/"
)
model = "claude-haiku-5.5"
```

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages= [
    {
        "role": "user",
        "content": "How does decorator pattern work in Python?"
    }
    ]
)
```

```python
for block in message.content:
    if block.type == "text":
        print(block.text)
```
